```python
import random

class VacuumEnvironment:
    """Two-square world: locations 'A' and 'B', each Clean (0) or Dirty (1)."""
    def __init__(self):
        self.status = {'A': random.randint(0, 1), 'B': random.randint(0, 1)}
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'

def reflex_vacuum_agent(percept):
    """Condition-action rules: decide using ONLY the current percept."""
    location, dirty = percept
    if dirty:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'

random.seed(7)
env = VacuumEnvironment()
env.status = {'A': 1, 'B': 1}; env.agent_location = 'B'
print("Initial state :", env.status, "| Agent at", env.agent_location)
score = 0
for step in range(1, 5):
    p = env.percept()
    action = reflex_vacuum_agent(p)
    env.execute(action)
    if action == 'Suck':
        score += 10      # reward for cleaning
    else:
        score -= 1       # cost of moving
    print(f"Step {step}: percept={p} -> action={action:5s} | world={env.status}")
print("Final state   :", env.status)
print("Performance score:", score)
```

```python
import random

class VacuumEnvironment:
    def __init__(self):
        self.status = {
            'A': random.randint(0, 1),
            'B': random.randint(0, 1)
        }
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location,
                self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'
        elif action == 'NoOp':
            pass


def reflex_agent(percept):
    location, dirty = percept

    if dirty == 1:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'


class ModelBasedAgent:
    def __init__(self):
        self.model = {'A': None, 'B': None}

    def act(self, percept):
        location, dirty = percept

        # Remember the current square's condition
        self.model[location] = dirty

        if dirty == 1:
            self.model[location] = 0
            return 'Suck'

        # Stop when both squares are clean
        if self.model['A'] == 0 and self.model['B'] == 0:
            return 'NoOp'

        # Visit the other square
        if location == 'A':
            return 'Right'
        else:
            return 'Left'


def run_agent(agent, initial_status, initial_location,
              steps=20):

    env = VacuumEnvironment()
    env.status = initial_status.copy()
    env.agent_location = initial_location

    score = 0

    for step in range(steps):
        percept = env.percept()

        if isinstance(agent, ModelBasedAgent):
            action = agent.act(percept)
        else:
            action = agent(percept)

        if action == 'Suck':
            if env.status[env.agent_location] == 1:
                score += 10
        elif action in ['Left', 'Right']:
            score -= 1

        env.execute(action)

    return score


# Use the same initial state for both agents
initial_status = {'A': 1, 'B': 1}
initial_location = 'B'

reflex_score = run_agent(
    reflex_agent, initial_status, initial_location
)

model_agent = ModelBasedAgent()
model_score = run_agent(
    model_agent, initial_status, initial_location
)

print("Initial state:", initial_status)
print("Agent location:", initial_location)
print("Simple Reflex Agent Score:", reflex_score)
print("Model-Based Agent Score:", model_score)
```

```python
import random

class VacuumEnvironment:
    """Two-square vacuum world: A and B."""

    def __init__(self, status=None, location=None):
        if status is None:
            status = {
                'A': random.randint(0, 1),
                'B': random.randint(0, 1)
            }

        self.status = status.copy()
        self.agent_location = (
            location if location is not None
            else random.choice(['A', 'B'])
        )

    def percept(self):
        return (self.agent_location,
                self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'
        elif action == 'NoOp':
            pass


# Simple reflex agent
def reflex_agent(percept):
    location, dirty = percept

    if dirty == 1:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'


# Random agent
def random_agent(percept):
    return random.choice(['Left', 'Right', 'Suck'])


# Model-based agent
class ModelBasedAgent:
    def __init__(self):
        self.memory = {'A': None, 'B': None}

    def act(self, percept):
        location, dirty = percept

        # Remember the current location's condition
        self.memory[location] = dirty

        # Stop when both locations are clean
        if self.memory['A'] == 0 and self.memory['B'] == 0:
            return 'NoOp'

        # Clean the current location if it is dirty
        if dirty == 1:
            return 'Suck'

        # Visit the other location
        if location == 'A':
            return 'Right'
        else:
            return 'Left'


# Run an agent for a fixed number of steps
def run_agent(agent, initial_status, initial_location,
              steps=20):

    env = VacuumEnvironment(
        status=initial_status,
        location=initial_location
    )

    score = 0

    for step in range(steps):
        percept = env.percept()

        # Choose an action
        if isinstance(agent, ModelBasedAgent):
            action = agent.act(percept)
        else:
            action = agent(percept)

        # Reward cleaning only when the square is dirty
        if action == 'Suck':
            if env.status[env.agent_location] == 1:
                score += 10

        # Penalize movement
        elif action in ['Left', 'Right']:
            score -= 1

        env.execute(action)

    return score, env.status


# ---------------------------------------
# TASK 3: RANDOM AGENT VS REFLEX AGENT
# ---------------------------------------

random.seed(7)

reflex_scores = []
random_scores = []

print("TASK 3: REFLEX AGENT VS RANDOM AGENT")
print("-" * 55)

for trial in range(1, 11):

    # Use the same initial state for both agents
    initial_status = {
        'A': random.randint(0, 1),
        'B': random.randint(0, 1)
    }
    initial_location = random.choice(['A', 'B'])

    score1, _ = run_agent(
        reflex_agent, initial_status, initial_location
    )

    score2, _ = run_agent(
        random_agent, initial_status, initial_location
    )

    reflex_scores.append(score1)
    random_scores.append(score2)

    print(f"Trial {trial}: Reflex = {score1:3d}, "
          f"Random = {score2:3d}")

print("\nAverage Reflex Score:",
      sum(reflex_scores) / len(reflex_scores))

print("Average Random Score:",
      sum(random_scores) / len(random_scores))


# ---------------------------------------
# TASK 4: MODEL-BASED VS REFLEX AGENT
# ---------------------------------------

model_scores = []
reflex_scores_2 = []

print("\n\nTASK 4: MODEL-BASED AGENT VS REFLEX AGENT")
print("-" * 55)

for trial in range(1, 11):

    # Use the same initial state for both agents
    initial_status = {
        'A': random.randint(0, 1),
        'B': random.randint(0, 1)
    }
    initial_location = random.choice(['A', 'B'])

    score1, _ = run_agent(
        reflex_agent, initial_status, initial_location
    )

    model = ModelBasedAgent()

    score2, _ = run_agent(
        model, initial_status, initial_location
    )

    reflex_scores_2.append(score1)
    model_scores.append(score2)

    print(f"Trial {trial}: Reflex = {score1:3d}, "
          f"Model-Based = {score2:3d}")

average_reflex = sum(reflex_scores_2) / 10
average_model = sum(model_scores) / 10

print("\nAverage Reflex Score:", average_reflex)
print("Average Model-Based Score:", average_model)

if average_model > average_reflex:
    print("Result: Model-based agent performed better.")
elif average_model == average_reflex:
    print("Result: Both agents had the same average score.")
else:
    print("Result: Reflex agent performed better.")
```
